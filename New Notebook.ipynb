```python
import arcpy
import os

arcpy.env.overwriteOutput = True

project_folder = r"C:\Users\hveisi\OneDrive - UWSP\Documents\ArcGIS\Projects\Erie_CSO_EJ_Analysis"
gdb = os.path.join(project_folder, "Erie_CSO_EJ_Analysis.gdb")
arcpy.env.workspace = gdb

cso_points = os.path.join(gdb, "cso_points_clean")
tracts = os.path.join(gdb, "tracts_EJ_clean")

buffer_1km = os.path.join(gdb, "model_cso_exposure_zone_1km")
exposed_tracts = os.path.join(gdb, "model_tracts_exposed_1km")
non_exposed_tracts = os.path.join(gdb, "model_tracts_non_exposed_1km")

print("Workspace:", gdb)

# 1. Create 1 km exposure buffer
arcpy.analysis.Buffer(
    in_features=cso_points,
    out_feature_class=buffer_1km,
    buffer_distance_or_field="1 Kilometers",
    dissolve_option="ALL"
)

print("Buffer created.")

# 2. Make tract layer
tract_layer = "tract_layer"
arcpy.management.MakeFeatureLayer(tracts, tract_layer)

# 3. Select exposed tracts
arcpy.management.SelectLayerByLocation(
    in_layer=tract_layer,
    overlap_type="INTERSECT",
    select_features=buffer_1km,
    selection_type="NEW_SELECTION"
)

arcpy.management.CopyFeatures(tract_layer, exposed_tracts)
print("Exposed tracts created.")

# 4. Select non-exposed tracts
arcpy.management.SelectLayerByLocation(
    in_layer=tract_layer,
    overlap_type="INTERSECT",
    select_features=buffer_1km,
    selection_type="NEW_SELECTION",
    invert_spatial_relationship="INVERT"
)

arcpy.management.CopyFeatures(tract_layer, non_exposed_tracts)
print("Non-exposed tracts created.")

print("Model completed.")
```
